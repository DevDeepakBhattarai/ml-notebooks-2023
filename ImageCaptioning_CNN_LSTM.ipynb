```python
folder_path='/content/flickr30k_images/flickr30k_images'
```

```python
import pandas as pd
unprocessed_df=pd.read_csv('/content/flickr30k_images/results.csv',delimiter='|')
unprocessed_df.drop(columns=' comment_number', inplace=True)
unprocessed_df[' comment'] = unprocessed_df[' comment'].apply(lambda x: str(x) if type(x) != str else x)
print(unprocessed_df.shape)
unprocessed_df.head()
```

```python
train_y=unprocessed_df[" comment"].values
print(train_y)
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts(train_y)
```

```python
print(len(tokenizer.word_index))
vocab_size=len(tokenizer.word_index) +1
tokenizer.word_index
```

```python
tokenized_sentences=tokenizer.texts_to_sequences(train_y)
tokenized_sentences
```

```python
max_comment_length=max([len(sentence) for sentence in tokenized_sentences])
max_comment_length
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_sequences=pad_sequences(tokenized_sentences,maxlen=max_comment_length,padding='post')
sequence_df=pd.DataFrame({'text':[seq.tolist() for seq in padded_sequences]})
padded_sequences.shape
```

```python
df=pd.concat([unprocessed_df['image_name'],sequence_df],axis=1)
df.head()
```

```python
import math
valid_size =int(math.floor(0.2 * 158915))
test_size= int(math.floor(0.1 * 158915))
sampled_df=df.sample(frac=1, random_state=0)
train_df = sampled_df.iloc[ : 158915 - valid_size]
valid_df = sampled_df.iloc[158915- valid_size :158915 - test_size ]
test_df = sampled_df.iloc[158915 -test_size : ]
```

```python
train_df.head()
# valid_df.head()
test_df.head()
```

```python
from tensorflow.keras.layers import Dense , LSTM , Conv2D, MaxPool2D ,Flatten, Embedding ,RepeatVector
from tensorflow.keras.models import Sequential, Model
import numpy as np
```

```python
import tensorflow as tf
from PIL import Image
def generate_batches(dataframe, batch_size):
    image_names = dataframe['image_name'].values
    texts = np.array(dataframe['text'].values.tolist())

    num_samples = len(dataframe)
    steps_per_epoch = num_samples // batch_size

    for step in range(steps_per_epoch):
        start = step * batch_size
        end = (step + 1) * batch_size

        batch_image_name = image_names[start:end]
        batch_texts = texts[start:end]

        batch_images = np.array([ np.array(Image.open(folder_path + "/"+ image_name).convert('RGB').resize((256, 256))) for image_name in batch_image_name])

        yield batch_images, batch_texts

batch_size=100
train_data_generator = generate_batches(train_df, batch_size)

train_dataset = tf.data.Dataset.from_generator(
    lambda: train_data_generator,
    output_signature=(
        tf.TensorSpec(shape=(batch_size, 256, 256, 3), dtype=tf.uint8),  # Assuming images are loaded as uint8 tensors
        tf.TensorSpec(shape=(batch_size,max_comment_length), dtype=tf.int32)
    )
)
valid_data_generator = generate_batches(valid_df, batch_size)

valid_dataset = tf.data.Dataset.from_generator(
    lambda: valid_data_generator,
    output_signature=(
        tf.TensorSpec(shape=(batch_size, 256, 256, 3), dtype=tf.uint8),  # Assuming images are loaded as uint8 tensors
        tf.TensorSpec(shape=(batch_size,max_comment_length), dtype=tf.int32)
    )
)
test_data_generator = generate_batches(test_df, batch_size)

test_dataset = tf.data.Dataset.from_generator(
    lambda: test_data_generator,
    output_signature=(
        tf.TensorSpec(shape=(batch_size, 256, 256, 3), dtype=tf.uint8),  # Assuming images are loaded as uint8 tensors
        tf.TensorSpec(shape=(batch_size,max_comment_length), dtype=tf.int32)
    )
)

valid_dataset=valid_dataset.repeat()
train_dataset=train_dataset.repeat()
```

```python
model=Sequential()

model.add(Conv2D(128,kernel_size=(3,3),strides=(1,1),input_shape=(256,256,3), padding='valid'))
model.add(MaxPool2D(pool_size=(2,2),strides=(2,2),padding="valid"))

model.add(Conv2D(64,kernel_size=(3,3),strides=(1,1), padding='valid'))
model.add(MaxPool2D(pool_size=(2,2),strides=(2,2),padding="valid"))

model.add(Conv2D(64,kernel_size=(3,3),strides=(1,1), padding='valid'))
model.add(MaxPool2D(pool_size=(2,2),strides=(2,2),padding="valid"))

model.add(Flatten())
model.add(Dense(128,activation="relu"))

model.add(RepeatVector(max_comment_length))

model.add(LSTM(256,return_sequences=True))

model.add(Dense(vocab_size ,activation='softmax'))

model.summary()
```

```python
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy",metrics=['accuracy'])
```
